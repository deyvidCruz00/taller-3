#✅**1. Numpy**

NumPy es una librería de Python especializada en el cálculo numérico y el análisis de datos, especialmente para un gran volumen de datos.

Incorpora una nueva clase de objetos llamados arrays que permite representar colecciones de datos de un mismo tipo en varias dimensiones, y funciones muy eficientes para su manipulación.

La ventaja de Numpy frente a las listas predefinidas en Python es que el procesamiento de los arrays se realiza mucho más rápido (hasta 50 veces más) que las listas, lo cual la hace ideal para el procesamiento de vectores y matrices de grandes dimensiones.

###**Operaciones matemáticas con arrays**

Existen dos formas de realizar operaciones matemáticas con arrays: a nivel de elemento y a nivel de array.

Las operaciones a nivel de elemento operan los elementos que ocupan la misma posición en dos arrays. Se necesitan, por tanto, dos arrays con las mismas dimensiones y el resultado es una array de la misma dimensión.

Los operadores mamemáticos +, -, *, /, %, ** se utilizan para la realizar suma, resta, producto, cociente, resto y potencia a nivel de elemento.

```python
a = np.array([[1, 2, 3], [4, 5, 6]],dtype=np.int32)
b = np.array([[1, 1, 1], [2, 2, 2]])
print(a)
print(b)
print("--------")
print(a + b )
print("--------")
print(a ** 2)
print("--------")
print(a*b)
```

```python
a_float=a.astype(np.float64)  # equivalente a un casting
print(a_float)
arr=np.array([3.7,-1.2,-2.6,0.5,12.9,10.1])
print(arr)
print("-------")
arr=arr.astype(np.int32)
print(arr)
print("-------")
numeric_strings=np.array(['1.24','-9.4','42'])
print(numeric_strings.astype(np.float64))
```

###**Inicialización de arrays**

```python
zeros = np.zeros(5)
print(zeros.dtype)
print(zeros)
ones = np.ones((4, 3))
print("--------")
print(ones)
identity = np.identity(4)
print("--------")
print(identity)
print("--------")
print(np.eye(4))
```

###**Acceso a los elementos de un array**

```python
a = np.array([[1, 2, 3], [4, 5, 6]])
print(a[1,0])  # Acceso al elemento de la fila 1 columna 0
print("--------")
print(a[1][0])  # Otra forma de acceder al mismo elemento
print("--------")
print(a[1:,0])
print("--------")
print(a[:, 0:2])
```

###**Filtrado de elementos de un array**

Una característica muy útil de los arrays es que es muy fácil obtener otro array con los elementos que cumplen una condición.

a[condicion] : Devuelve un arreglo de **una** dimensión con los elementos del array que cumplen la condición.

```python
a = np.array([[1, 2, 3], [4, 5, 6]])
print(a)
print("--------")
print(a[(a % 2 == 0)])
print("--------")
print(a[(a % 2 == 0) &  (a > 2)])
```

```python
print(a[(a>=1)])
```

###**np.linspace(inicio, fin, n)**

Crea y devuelve una referencia a un array de una dimensión cuyos elementos son la secuencia de n valores equidistantes desde inicio hasta fin.

**Diferencia entre np.linspace and np.arange --->** Con np.arange() se define cada cuánto se desea que genere los elementos y con ello dará un tamaño de array. En cambio con np.linspace() se define el tamaño del array (num) y los steps (pasos) se calcularán automáticamente.

```python
import numpy as np
np.linspace(start=-5, stop=5, num=10)
```

```python
numbers = np.linspace(-10, 10, 11, dtype=int)
print(numbers)
print(numbers.dtype)
```

```python
np.linspace(1, 5, 100)
```

```python
ytemperatures=np.linspace(13.6, 18.2, 20)
ytemperatures
```

```python
arr_slice[1] = 12345
print(arr)
print(arr_slice)

arr_slice[:] = 64
print(arr)
print(arr_slice)
```
