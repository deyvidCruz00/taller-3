#✅**2. Pandas**

Proporciona estructuras de datos de alto nivel y funciones diseñadas para funcionar con datos estructurados o tabulares rápidos, fáciles y expresivos. Es un entorno de análisis de datos potente y productivo. Las estructuras de datos en pandas son:

* Serie, arreglo de objetos unidimensional.
* DataFrame, estructura de datos tabular orientada a columnas con etiquetas de fila y columna.

Principales características de Pandas:
* Combina las ideas de alto rendimiento y computación en matrices de NumPy con la flexibilidad y capacidades de manipulación de datos de hojas de cálculo y bases de datos relacionales (SQL).
* Proporciona una sofisticada funcionalidad de indexación para facilitar manejo de dados, realizar agregaciones y seleccionar subconjuntos de datos.
* Gestión de tareas ETL para limpieza con mucha habilidad en el análisis de datos.
*Permite leer y escribir fácilmente archivos en formato CSV, Excel y bases de datos SQL.
* Funcionalidad integrada de series de tiempo. Las mismas estructuras de datos manejan tanto datos de series temporales como datos de series no temporales.
* Operaciones aritméticas y reducciones que conservan los metadatos.
* Manejo flexible de datos faltantes.
*Fusionar y otras operaciones relacionales encontradas en bases de datos populares (basadas en SQL)

🌵**DataFrame**

Un DataFrame es una estructura de datos que almacena la información como una tabla ordenada por filas y columnas. Cada fila representa un objeto y cada columna la información correspondiente a una característica de los objetos.

Un DataFrame también posee índices por cada fila, que pueden ser dados o generados automáticamente. Cada columna del DataFrame es una [**serie**], donde el valor del índice corresponde con los valores de índice que tiene el DataFrame.


Por medio de un diccionario vamos a crear un dataframe, donde las llaves son los nombres de las columnas y los valores son las características.

Por ejemplo, para hacer un DataFrame con el equipo de fútbol anterior, pero agregando las series de estatura y peso, se hará de la siguiente manera:

```python
import pandas as pd
dict_caracteristicas = {'apellido':['Ospina', 'Zapata', 'Falcao', 'Cuadrado', 'Rodriguez'],
                       'altura':[183.0,187.0,177.0,179.0,180.0],
                       'peso':[80.0,82.0,72.0,72.0,75.0]}

seleccionColombia = pd.DataFrame(dict_caracteristicas,index=[1, 2, 9, 11, 10])
```

Al imprimir el DataFrame, se observa que su estructura es similar a la de un documento en Excel, donde el índice (que no tiene nombre de columna) es el número de la camiseta del jugador.

```python
seleccionColombia
```

**Acceso a los registros del DataFrame por el índice**

En pandas podemos acceder a la fila por el índice de ésta. En el ejemplo anterior éste índice era el número de la camiseta del jugador. Pero podría ser una fecha, una letra, etc.

Si conocemos el índice simplemente lo usamos para ir directamente a la fila deseada.

```python
seleccionColombia['apellido']
```

```python
seleccionColombia[seleccionColombia.apellido=='Ospina']
```

Para **agregar** un nuevo jugador (**FILA**), se puede utilizar el índice, que en este caso es el número de la camiseta en el equipo y un arreglo representando todas sus características (apellido, altura y peso).

NOTA: Las características tienen que estar en el mismo orden.

Imprimiendo el DataFrame podemos ver el nuevo jugador.

```python
num_jugador = 3
caract_jugador = ['Murillo',184,80.0]
seleccionColombia.loc[num_jugador] = caract_jugador
```

```python
seleccionColombia
```

Para **añadir** una nueva una **serie o columna** a un Dataframe, por ejemplo la masa corporal

```python
masa=[71,73,80,70,68,81]
seleccionColombia['masa']=masa
seleccionColombia
```

# Concatenación de dataframes
https://aprendeconalf.es/docencia/python/manual/pandas/#:~:text=Para%20a%C3%B1adir%20una%20fila%20a,de%20las%20columnas%20de%20df%20.

```python
# Concatenación de Dataframes por FILAS
import pandas as pd
df1 = pd.DataFrame({"Nombre":["Carmen", "Luis"],"Sexo":["Mujer", "Hombre"], "Edad":[22, 18]}).set_index("Nombre")
print('-----------')
print(df1)
df2 = pd.DataFrame({"Nombre":["María", "Pedro"],"Sexo":["Mujer", "Hombre"], "Edad":[25, 30]}).set_index("Nombre")
print('-----------')
print(df2)
print('---------')
df = pd.concat([df1, df2])
print(df)
```

```python
# Concatenación de Dataframes por COLUMNAS
df1 = pd.DataFrame({"Nombre":["Carmen", "Luis", "María"],"Sexo":["Mujer", "Hombre", "Mujer"]}).set_index("Nombre")
print('-----------')
print(df1)
df2 = pd.DataFrame({"Nombre":["Carmen", "Luis", "María"],"Edad":[22, 18, 25]}).set_index("Nombre")
print('-----------')
print(df2)
df = pd.concat([df1, df2], axis = 1)
print('-----------')
print(df)
```

# Mezcla de dataframes

```python
import pandas as pd
df1 = pd.DataFrame({"Nombre":["Carmen", "Luis", "María"],  "Sexo":["Mujer", "Hombre", "Mujer"]})
print('-----------')
print(df1)
df2 = pd.DataFrame({"Nombre":["María", "Pedro", "Luis"], "Edad":[25, 30, 18]})
print('-----------')
print(df2)
#df = pd.merge(df1, df2, on="Nombre")
df = pd.merge(df1, df2, on="Nombre", how="outer")
print('----------')
print(df)
```

# None and NaN

In Pandas missing data is represented by two value: None: None is a Python singleton object that is often used for missing data in Python code. NaN : NaN (an acronym for Not a Number),

```python
import numpy as np
```

```python
df = pd.DataFrame({'col1':[1,2,3,np.nan],
                   'col2':[np.nan,555,666,444],
                   'col3':['abc','def','ghi',None]})
df.head()
```

```python
df.fillna('0', inplace=True)
```

El método **fillna** permite sustituir los valores nulos de una estructura pandas por otro valor según ciertos criterios.

El método **dropna** permite, de una forma muy conveniente, filtrar los valores de una estructura de datos pandas para dejar solo aquellos no nulos.
