#🌵 Consumir datos de archivos con Formato Json, ubicado en la sesión

Aquí encontrará el archivo [_personFile.json](https://drive.google.com/file/d/1CF_XSkXW_dBcO3FMZspC2x6tyqxU4pjJ/view?usp=sharing)

Luego de cargado el archivo de JSON hacia un DataFrame, las operaciones de gestion, indexacción y acceso son equivalentes a lo visto en el anterior formato.

```python
import json
with open("_personFile.json","r") as j:
  mydata = json.load(j)
```

```python
print(type(mydata))
```

```python
df_person=pd.DataFrame(mydata)
```

```python
print(type(df_person))
```

```python
df_person.dtypes
```

```python
df_person.head()
```

```python
list_tags=df_person['tags'] # palabras reservadas o keywords
list_tags
```

```python
list_tags_1=df_person.loc[1:1,['tags']]
print(type(list_tags_1))
print(list_tags_1)
print(list_tags_1['tags'])
print(list_tags_1['tags'][1])
print("------")
npArr=list_tags_1.to_numpy()
print(type(npArr))
print(npArr)
print(npArr[0])
print(npArr[0][0])
print(npArr[0][0][0])
```

```python
# https://www.delftstack.com/es/howto/python-pandas/pandas-column-to-list/
# Una columna en el DataFrame de Pandas es una Serie.
# Para convertir una columna en una lista, podemos usar el método tolist().
# tolist() convierte la Serie del Dataframe en una lista.
list_of_single_column = df_person['tags'].tolist()
print(type(list_of_single_column))
print(list_of_single_column)
print(list_of_single_column[1])
print(list_of_single_column[1][0])
```
